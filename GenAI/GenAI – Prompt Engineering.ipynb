### Task 1: Replace Hardcoded Prompts

```python
# Import first
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate

# PromptTemplate version (Reusable)
template_1 = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

def explain_topic(topic):
    return template_1.format(topic=topic)

# Test
print(explain_topic("AI"))
```

### Task 2: Multi-Input Prompt System

```python
template_2 = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

# Test Cases
print(template_2.format(topic="AI", audience="beginners", tone="friendly"))
print(template_2.format(topic="Python", audience="kids", tone="fun"))
print(template_2.format(topic="Deep Learning", audience="engineers", tone="technical"))
```

### Task 3: Prompt Variations Engine

```python
# Teaching Template
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

# Interview Template
interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 questions about {topic}"
)

# Storytelling Template
story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

topic = "Machine Learning"

print(teaching_template.format(topic=topic))
print(interview_template.format(topic=topic))
print(story_template.format(topic=topic))
```

### Task 4: ChatPromptTemplate System

```python
def get_chat_prompt(role, topic):
    
    if role == "teacher":
        system_msg = "You are a helpful teacher who explains concepts clearly."
    elif role == "interviewer":
        system_msg = "You are an interviewer who asks challenging questions."
    elif role == "motivator":
        system_msg = "You motivate and inspire people."
    else:
        system_msg = "You are a helpful assistant."
    
    chat_template = ChatPromptTemplate.from_messages([
        ("system", system_msg),
        ("user", "Explain {topic}")
    ])
    
    return chat_template.format_messages(topic=topic)

# Test
messages = get_chat_prompt("teacher", "Neural Networks")
print(messages)
```

### Task 5: Input Validation Layer

```python
def validate_inputs(audience, tone):
    
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]
    
    if audience not in valid_audience:
        print("Invalid audience! Defaulting to 'beginner'")
        audience = "beginner"
        
    if tone not in valid_tone:
        print("Invalid tone! Defaulting to 'formal'")
        tone = "formal"
        
    return audience, tone

# Test
print(validate_inputs("kids", "cool"))
```

### Task 6: Prompt Generator App

```python
def generate_prompt(topic, audience, tone, style):
    
    audience, tone = validate_inputs(audience, tone)
    
    templates = {
        "teaching": "Explain {topic} for {audience} in a {tone} teaching style",
        "interview": "Ask interview questions on {topic} for {audience} in a {tone} tone",
        "storytelling": "Explain {topic} for {audience} in a {tone} storytelling style"
    }
    
    if style not in templates:
        raise ValueError("Invalid style")
    
    prompt_template = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template=templates[style]
    )
    
    return prompt_template.format(
        topic=topic,
        audience=audience,
        tone=tone
    )

# Test
print(generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))
```

### Task 7: Template Reusability Test

```python
reusable_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

inputs = [
    ("AI", "beginner", "fun"),
    ("Python", "intermediate", "formal"),
    ("NLP", "expert", "casual"),
    ("Data Science", "beginner", "formal"),
    ("Deep Learning", "expert", "fun")
]

for t, a, to in inputs:
    print(reusable_template.format(topic=t, audience=a, tone=to))
```

### Pipeline Flow

User Input → Validation → Prompt Template → Dynamic Prompt Generation → Output
