## 1. Smart Parking Lot Management System

```python
def smart_parking_system(capacity, vehicle_logs):
    current_count = 0
    peak_count = 0
    for log in vehicle_logs:
        if log == "IN":
            current_count += 1
        elif log == "OUT" and current_count > 0:
            current_count -= 1
        peak_count = max(peak_count, current_count)
    
    if current_count > capacity:
        status = "ALERT: Parking Full"
    else:
        status = "Available"

    print(f"Currently Parked Vehicles: {current_count}")
    print(f"Parking Status: {status}")

capacity= 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
smart_parking_system(capacity, vehicle_logs)
```

## 2. Online Food Delivery Time Estimator

```python
def delivery_time(distance,traffic,weather):
    # Base time (5 minutes per km)
    time = distance * 5
    
    # Traffic delay
    if traffic == 'Medium':
        time += 5
    elif traffic == 'High':
        time += 10
    
    # Weather delay
    if weather == 'Rainy':
        time += 5
    print(f"Estimated Delivery Time: {time} minutes")
    
distance = 8
traffic = "High"
weather = "Rainy"

delivery_time(distance, traffic, weather)
```

## 3.Movie Theatre Seat Occupancy Analyzer

```python
def analyze_seat(total_seats,booked_seats):
    booked_count = len(booked_seats)
    occupancy = (booked_count/total_seats)*100
    if occupancy == 100:
        status = "Housefull"
        suggestion = "Open Additional Show"
    elif occupancy >= 75:
        status = "Almost Full"
        suggestion = "Consider Opening Additional Show"
    else:
        status = "Seats Available"
        suggestion = "No Additional Show Needed"

    print(f"Occupancy: {int(occupancy)}%")
    print(f"Show Status: {status}")

total_seats= 200
booked_seats= [1]*150
analyze_seat(total_seats, booked_seats)
```

## 4. Cloud Server Load Classification System

```python
def server_load(readings):
    total = 0
    for reading in readings:
        total += reading
    average = total/len(readings)
    
    if average > 80:
        print(f"Average CPU Load: {average}%\nServer Status: Critical")
    elif 80 >= average > 50:
        print(f"Average CPU Load: {average}%\nServer Status: Warning")
    else:
        print(f"Average CPU Load: {average}%\nServer Status: Normal")

readings = [45, 60, 70, 85, 90]
server_load(readings)
    
```

## 5. Smart Classroom Resource Usage Monitor

```python
def monitor_resources(resource_usage):
    overused = []

    for resource, hours in resource_usage.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        alert = "Yes"
    else:
        alert = "No"

    print("Overused Resources:", ", ".join(overused) if overused else "None")
    print("Energy Alert:", alert)
    
resource_usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

monitor_resources(resource_usage)
```

## 6. Online Event Registration Capacity Controller

```python
def event_registration_controller(capacity, registrations):
    confirmed = 0
    waitlisted = 0
    for i in range(registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1
    status = "Closed" if confirmed == capacity else "Open"

    print(f"Confirmed Registrations: {confirmed}")
    print(f"Waitlisted Users: {waitlisted}")
    print(f"Registration Status: {status}")
    
capacity = 100
registrations = 105
event_registration_controller(capacity, registrations)
```
