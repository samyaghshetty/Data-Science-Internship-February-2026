## 1. Smart Light Controller

```python
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"
    def turn_on(self):
        self.status = "ON"
    def turn_off(self):
        self.status = "OFF"
    def display_status(self):
        print(f"{self.name} is {self.status}")
light_name = input("Light Name:").strip()
action = input("Action:").strip().upper()
light = SmartLight(light_name)
if action == "ON":
    light.turn_on()
elif action == "OFF":
    light.turn_off()
light.display_status()
```

## 2. Employee ID Card System

```python
class Employee:
    def __init__(self,name,e_id,department):
        self.name = name
        self.e_id = e_id
        self.department = department
    def display(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.e_id}")
        print(f"Department: {self.department}")       
employe = Employee("Rahul","EMP102","IT")
employe.display()
```

## 3. Mobile Contact Record

```python
class Contact:
    def __init__(self,name,phone):
        self.name = name
        self.phone = phone
    def save_contact(self):
        print("Contact Saved")
    def show_contact(self):
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")
contact = Contact("Anita", "9876543210")
contact.save_contact()
contact.show_contact()
```

## 4. Product Price Tag Generator

```python
class Product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def show_product(self):
        print(f"Name: {self.name}")
        print(f"Price: {self.price}")
product = Product("Headphones","₹2499")
product.show_product()
```

## 5. Movie Rating Display System

```python
class Movie:
    def __init__(self,name,rating):
        self.name = name
        self.rating = rating
    def show_details(self):
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating}/5")
movie = Movie("Inception","4.8")
movie.show_details()
```

## 6. Delivery Address Manager

```python
class Delivery:
    def __init__(self, name, address):
        self.name = name
        self.address = address
    def show_details(self):
        print("Delivery Details")
        print(f"Customer: {self.name}")
        print(f"Address: {self.address}")
delivery = Delivery("Suman", "Hyderabad")
delivery.show_details()
```
